# Ejemplo básico de análisis de datos con Python

Este notebook utiliza un pequeño conjunto de datos para calcular estadísticas simples.

```python
import pandas as pd

# Crear un pequeño conjunto de datos
datos = {
    'horas_estudio': [2, 3, 4, 5, 6],
    'nota': [4.0, 4.8, 5.5, 6.2, 6.8]
}

df = pd.DataFrame(datos)
df
```

## Promedio

Calculamos el promedio de las horas de estudio y de las notas.

```python
promedio_horas = df['horas_estudio'].mean()
promedio_nota = df['nota'].mean()

print(f'Promedio de horas de estudio: {promedio_horas:.2f}')
print(f'Promedio de nota: {promedio_nota:.2f}')
```

## Relación entre horas de estudio y nota

Creamos un gráfico sencillo para observar si existe una relación entre ambas variables.

```python
import matplotlib.pyplot as plt

plt.scatter(df['horas_estudio'], df['nota'])
plt.xlabel('Horas de estudio')
plt.ylabel('Nota')
plt.title('Horas de estudio vs. nota')
plt.show()
```
